# Test Transform Functions

This notebook tests all transformation functions from `utils/transforms.py`


```python
import sys
from pathlib import Path

# Add parent directory to path
sys.path.append(str(Path().resolve().parent))

import pandas as pd
from datetime import datetime
from utils.transforms import (
    load_clean_data,
    filter_by_date_range,
    filter_by_category,
    filter_by_label,
    get_current_month_expenses,
    get_last_month_expenses,
    get_expenses_by_category,
    get_expenses_by_month,
    get_top_transactions
)
```

```python
# Load clean data
df = load_clean_data()
print(f"DataFrame shape: {df.shape}")
print(f"\nColumns: {df.columns.tolist()}")
print(f"\nDate range: {df['date'].min()} to {df['date'].max()}")
df.head()
```

## Test Filter Functions


```python
# Test filter_by_date_range
start_date = pd.Timestamp("2025-01-01")
end_date = pd.Timestamp("2025-01-31")
filtered = filter_by_date_range(df, start_date, end_date)
print(f"Filtered by date range: {len(filtered)} rows")
filtered.head()
```

```python
# Test filter_by_category
categories = ["Groceries", "Restaurant"]
filtered = filter_by_category(df, categories)
print(f"Filtered by category: {len(filtered)} rows")
print(f"Unique categories: {filtered['category'].unique()}")
filtered.head()
```

```python
# Test filter_by_label
labels = ["breakfast", "lunch"]
filtered = filter_by_label(df, labels)
print(f"Filtered by label: {len(filtered)} rows")
filtered.head()
```

## Test Get Data Functions


```python
# Test get_current_month_expenses
current_month = get_current_month_expenses(df)
print(f"Current month expenses: ${current_month:,.0f}")
```

```python
# Test get_last_month_expenses
last_month = get_last_month_expenses(df)
print(f"Last month expenses: ${last_month:,.0f}")
```

```python
# Test get_expenses_by_category
start_date = pd.Timestamp("2025-01-01")
end_date = pd.Timestamp("2025-01-31")
category_data = get_expenses_by_category(df, start_date, end_date)
print(f"Expenses by category: {len(category_data)} categories")
category_data.head()
```

```python
# Test get_expenses_by_month
monthly_data = get_expenses_by_month(df, year=2025)
print(f"Expenses by month: {len(monthly_data)} months")
monthly_data.head()
```

```python
# Test get_top_transactions
top_transactions = get_top_transactions(df, n=10, year=2025, month=1)
print(f"Top transactions: {len(top_transactions)} rows")
top_transactions.head()
```
